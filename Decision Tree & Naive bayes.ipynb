```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
RANDOM_STATE = 42
```

```python
df = load_iris()
```

```python
x = df.data
y = df.target
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=RANDOM_STATE)
```

```python
DT = DecisionTreeClassifier(criterion="gini", max_depth=3)  # max_depth controls the depth of the tree
```

```python
DT = DT.fit(x_train,y_train)
```

```python
y_pred = DT.predict(x_test)
```

```python
accuracy = metrics.accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

```python
# Additional: Visualize the Decision Tree (requires graphviz library)
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
# Plot the tree structure
plt.figure(figsize=(12,8))
plot_tree(DT, filled=True, feature_names=df.feature_names, class_names=df.target_names, rounded=True)
plt.show()
```

# GaussianNB

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
iris = load_iris()
```

```python
x = iris.data 
y = iris.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=RANDOM_STATE)
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train, y_train)
```

```python
y_pred = gnb.predict(X_test)
```

```python
accuracy = metrics.accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

```python
# Additional: Confusion matrix and classification report
conf_matrix = metrics.confusion_matrix(y_test, y_pred)
class_report = metrics.classification_report(y_test, y_pred)
```

```python
print("Confusion Matrix:\n", conf_matrix)
print("\nClassification Report:\n", class_report)
```

```python
# Draw the confusion matrix as a heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=iris.target_names, yticklabels=iris.target_names)

# Add titles and labels
plt.title('Confusion Matrix')
plt.ylabel('Actual Class')
plt.xlabel('Predicted Class')

# Show the plot
plt.show()
```
